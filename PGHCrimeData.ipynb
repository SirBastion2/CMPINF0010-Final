```python
import pandas as pd

pd.options.display.max_rows = 90

pghcrime = pd.read_csv("PGHCrime.tsv", sep="\t")
pghcrime.head(5)
```

```python
pghc = dict()
for index, row in pghcrime.dropna(subset=['Neighborhood']).iterrows():
    neighborhood = row['Neighborhood']
    if neighborhood not in pghc:
        pghc[neighborhood] = {
            "Felonies": 0,
            "Misdemeanors": 0,
            "Infractions": 0
        }
    if "FELONY" in str(row['ArrestCharge_Felony_Misdemeanor_Description']):
        pghc[neighborhood]["Felonies"] += 1
    if "MISDEMEANOR" in str(row['ArrestCharge_Felony_Misdemeanor_Description']):
        pghc[neighborhood]["Misdemeanors"] += 1
    if "INFRACTION" in str(row['ArrestCharge_Felony_Misdemeanor_Description']):
        pghc[neighborhood]["Infractions"] += 1
pghall = pd.DataFrame.from_dict(pghc, orient='index')
pghall.head(10)
```

All the code below lets you sort by ascending or descending order.

Set ascending=True to display in ascending order.

You can change head() to a maximum of 90.

```python
pghfel = pghall.sort_values(by='Felonies', ascending=False)
pghfel.head(5)
```

```python
pghmis = pghall.sort_values(by='Misdemeanors', ascending=False)
pghmis.head(5)
```

```python
pghinfra = pghall.sort_values(by='Infractions', ascending=False)
pghinfra.head(5)
```
